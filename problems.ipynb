### Explain how you would represent 32-bit words.

I would represent each 32-bit word using a Python integer. Python integers do not automatically restrict a number to 32 bits as they are arbitrary-precision binary, to ensure that each word remains within the required 32-bit range, a bitwise mask can be used.

SHA-256 performs many operations on 32-bit words, using Python integers allows these operations to be performed easily.

The hexadecimal value of 0xFFFFFFFF represents 32 bits that are all set to 1. This can be used as a mask to ensure that values remain within the 32-bit rang.

```python
#Representing 32-bit words
word1 = 0x12345678
word2 = 0x11223344
word3 = 0x99887766

#Ensuring they remain within 32 bits
word1 = word1 & 0xFFFFFFFF
word2 = word2 & 0xFFFFFFFF
word3 = word3 & 0xFFFFFFFF

print(hex(word1))
print(hex(word2))
print(hex(word3))
```

### Explain how you would represent sequences of 32-bit words

I would represent a sequence of 32-bit words as a Python list. Each element in the list is a Python integer which will be used to represent one 32-bit word. A List works well because it can store multiple values in a specific order, this is important when using the SHA-256 as this processes 32-bit words in a defined sequence.

Similarly to singular 32-bit words, these values can be kept in a 32-bit range by applying the 0xFFFFFFFF bitwise mask to each word.

```python
word1 = 0x12345678
word2 = 0x11223344
word3 = 0x44556677

#ensuring each word remains within 32 bits
word1 = word1 & 0xFFFFFFFF
word2 = word2 & 0xFFFFFFFF
word3 = word3 & 0xFFFFFFFF

#putting sequence of words into a list
words = [word1, word2, word3]

#prints hexidecimal string
print(hex(word1))
print(hex(word2))
print(hex(word3))

print([hex(word) for word in words])
```

### Explain how you would represent input messages

E.g of input message = "hello"

I would represent an input message as a Python string, then convert the string into bytes using UTF-8 encoding. A String is appropriate for representing the message because it allows text to be stored in readable form.

SHA-256 operates on binery data so the message needs to be converted into a sequence of bytes before it can be processed.

```python
#representing input message as a String
message = "hello"

#converting the message into bytes 
message_bytes = message.encode("utf-8")

#printing the input String
print(message)

#prints the bytes object - indicated by the "b"
print(message_bytes)
```
