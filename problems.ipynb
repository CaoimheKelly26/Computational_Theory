### Explain how you would represent 32-bit words.

I would represent each 32-bit word using a Python integer. Python integers do not automatically restrict a number to 32 bits as they are arbitrary-precision binary, to ensure that each word remains within the required 32-bit range, a bitwise mask can be used.

SHA-256 performs many operations on 32-bit words, using Python integers allows these operations to be performed easily.

The hexadecimal value of 0xFFFFFFFF represents 32 bits that are all set to 1. This can be used as a mask to ensure that values remain within the 32-bit rang.

```python
#Representing 32-bit words
word1 = 0x12345678
word2 = 0x11223344
word3 = 0x99887766

#Ensuring they remain within 32 bits
word1 = word1 & 0xFFFFFFFF
word2 = word2 & 0xFFFFFFFF
word3 = word3 & 0xFFFFFFFF

print(hex(word1))
print(hex(word2))
print(hex(word3))
```

### Explain how you would represent sequences of 32-bit words

I would represent a sequence of 32-bit words as a Python list. Each element in the list is a Python integer which will be used to represent one 32-bit word. A List works well because it can store multiple values in a specific order, this is important when using the SHA-256 as this processes 32-bit words in a defined sequence.

Similarly to singular 32-bit words, these values can be kept in a 32-bit range by applying the 0xFFFFFFFF bitwise mask to each word.

```python
word1 = 0x12345678
word2 = 0x11223344
word3 = 0x44556677

#ensuring each word remains within 32 bits
word1 = word1 & 0xFFFFFFFF
word2 = word2 & 0xFFFFFFFF
word3 = word3 & 0xFFFFFFFF

#putting sequence of words into a list
words = [word1, word2, word3]

#prints hexidecimal string
print(hex(word1))
print(hex(word2))
print(hex(word3))

print([hex(word) for word in words])
```

### Explain how you would represent 512-bit message blocks

A 512-bit message block is a fixed-size block of data which is the equivalent to 64 bytes which is used in cryptographic hash functions. It is the the block size used by SHA-256. I would represent a message block using Python's bytes data type, as it stores binary data efficiently.

Each block can be divided into sixteen 32-bit words for processing by the SHA-256 algorithm. 

This size ensures that even a small change in the input will result in a significantly different hash, enhancing the security of the hash function.

This representation is appropriate because Python's bytes type allows the block to be stored as a fixed sequence of bytes, making it easy to access and convert sections of the block into 32-bit words.

```python
# createing a 512-bit message block
block = bytes(64)

# displauing the size of the block
print("Block size in bytes: ", len(block))
print("Block size in bits: ", len(block) * 8)

# dividing the block into sixteen 32-bit segmetns
words = [
    int.from_bytes(block[i:i+4], "big")
    for i in range(0, 64, 4)
]

#printing the 32-bit words in hexadecimal
print([hex(word) for word in words])
```

### Explain how you would represent the final 256-bit hash value

A 256-bit hash is a fixed-length, 256-bit(32-byte) output generated by a cryptographic hash function, commonly used to ensure data integrity and security.

For the final 256-bit SHA-256 hash value, i would represent it using Python's bytes data type. A SHA-256 hash contains 256 bits, which is equivalent to 32 bytes.

I would chose Python's bytes data type becayse bytes can store the complete hash value as binary data wihtout chaning its contents. It can also be easily converted into a hexadecimal string, making the final hash easier to read and compare.

```python
# representing a 256-bit hash value using bytes
hash_value = bytes(32)

# printing the hash value hexidemical
print(hash_value.hex())
# printing the hash value
print(hash_value)

# checking the size of the hash
print("Hash size in bytes: ", len(hash_value))
print("Hash size in bits: ", len(hash_value * 8))
```

# Explain how you would represent the big endian 

In a big-endian system, the most significant byte (MSB) is stored at teh lowest memory address. meaning the "big end" comes first. A 32-bit integer 0x12345678 would be stored in memory as follows:

0x12 is the most signficant byte, which will be placed at the lowest address, which is (00) followed by 0x34, 0x56 and 0x78

I would represent big-endian integers using Python's int data type and the int.from_bytes() method because SHA-256 processes data in big-endian order, and Python allows bytes to be converted into integers using this byte order.


```python
# representing four bytes of data
data = bytes([0x12, 0x34, 0x56, 0x78])
# non hexidecimal format
data1 = bytes([23, 89, 140, 230])

#converting the bytes into a big endian integer
big_endian = int.from_bytes(data, "big")
big_endian1 = int.from_bytes(data1, "big")

print(hex(big_endian))
print(hex(big_endian1))
```
