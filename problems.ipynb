#### Representing input, output and intermediate data in Python. 

### Input
In Python, Input data is represented as bytes.

e.g message = b"hello" -> Python will represent this as a sequence of bytes.

h   e   l   l   o 

68  65  6c  6c  6f  <- Hexadecimal

### Intermediate Data
SHA-256 does its calculations using 32-bit numbers. In Python, these are represented using the normal int type.

e.g.
 
a = 0x6a09e667

b = 0xbb67ae85

SHA-256 thinks of these as 32-bit numbers, Python just sees them as integers.

### Output
At the end, SHA-256 produces 256 bits.

e.g message = b"hello" -> The output Python bytes will be

h   e   l   l   o

2c  f2  4d  ba  5f

these can be stored in a variable which is then converted to hexidecimal format

hex_digest = digest.hex()

so instead of 32 bytes we are left wiht 64 characters:
